**Nested If Else Statements**

A nested if-else statement in Python is simply an if or if-else block placed inside another if or else block. It is used when you need to check multiple conditions that depend on one another (layered decision-making)

```python
username = "admin"
password = "secret_password"

if username == "admin":
    if password == "secret_password":
        print("Welcome back, Admin!")
    else:
        print("Incorrect password.")
else:
    print("Username not found.")
```

```python
# Break, Continue, Else, Pass
# The break statement is used to exit the loop prematurely when a certain condition is met.
# When the break statement is executed, Python terminates the loop, even if the while condition is true.

counter = 0

while counter <= 10:
  print(counter)
  if counter == 5:
    break
  counter += 1
```

```python
# Nested While Loops

number = 0

while number <= 10:
  print(number)
  number += 1
  while number <= 5:
    print(number)
    number += 1

```

```python
day = 1

# 1. Outer while loop (Tracks the days)
while day <= 3:
    print(f"--- Day {day} ---")

    log_count = 1
    # 2. Inner while loop (Tracks how many times we check temp per day)
    while log_count <= 2:
        print(f"  Taking temperature log #{log_count}...")

        log_count = log_count + 1  # Updates inner loop variable

    day = day + 1  # Updates outer loop variable
```

```python
outer_num = 1

# 1. Outer Loop: Controls the first number
while outer_num <= 3:
    inner_num = 1

    # 2. Inner Loop: Multiplies the outer number by 1, 2, and 3
    while inner_num <= 3:
        product = outer_num * inner_num
        print(f"{outer_num} x {inner_num} = {product}")

        inner_num = inner_num + 1  # Moves inner loop forward

    print("--- End of Row ---")  # Runs after the inner loop finishes
    outer_num = outer_num + 1    # Moves outer loop forward
```

```python
outer_num = 1

while outer_num <= 2:

    inner_num = 1
    while inner_num <= 5:
        print(outer_num, inner_num)
        inner_num = inner_num + 1  # count up inner

    outer_num = outer_num + 1      # count up outer
```

```python
# Normal(No List Comprehension )

tv_shows = ['friends', 'PARKS AND RECREATION', 'the office', 'big bang theory']

upper_tv_shows = []

for show in tv_shows:
  upper_tv_shows.append(show.upper())

print(upper_tv_shows)
```

```python
# List Comprehension
# https://youtu.be/DUnY6l482Lk

tv_shows = ['friends', 'PARKS AND RECREATION', 'the office', 'big bang theory']

shows_title = [show.upper() for show in tv_shows]

print(shows_title)
```

```python
# Let's try using List Comprehension to create a list from another list!
# Let's create a number checker that sees if a number is divisible by 2. If it is we can add it our list.
# Here is your starting list:

numbers = [2, 4, 6, 8, 10, 11, 13, 15]

new_numbers = [num for num in numbers if (num % 2) == 0]

print(new_numbers)
```
