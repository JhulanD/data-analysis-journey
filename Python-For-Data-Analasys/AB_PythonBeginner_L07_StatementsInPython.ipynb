**Nested If Else Statements**

A nested if-else statement in Python is simply an if or if-else block placed inside another if or else block. It is used when you need to check multiple conditions that depend on one another (layered decision-making)

```python
username = "admin"
password = "secret_password"

if username == "admin":
    if password == "secret_password":
        print("Welcome back, Admin!")
    else:
        print("Incorrect password.")
else:
    print("Username not found.")
```

```python
grade = 69


if grade >= 90:
  if grade >= 95:
    print('A+')
  elif grade >= 92:
    print('A')
  else:
    print('A-')
elif grade >= 80:
  if grade >= 85:
    print('B+')
  elif grade >= 82:
    print('B')
  else:
    print('B-')
elif grade >= 70:
  if grade >= 75:
    print('C+')
  elif grade >= 72:
    print('C')
  else:
    print('C-')
else:
  print('F')
#
```

### For Loops

```python
in_list = [1, 2, 3, 4, 5]

for number in in_list:
  print(number)
```

```python
for viper in in_list:
  print('Most Venomous')
```

```python
int_list = [1,2,3,4,5,6]

for num in int_list:
  power_of_num = num * num
  print(power_of_num)
```

```python
sale_amount = [1,10,20,75.75, 100,1000,80,50]

small_sale = []
medium_sale = []
large_sale = []

for sale in sale_amount:
  if sale < 10:
    small_sale.append(sale)
  elif sale < 75:
    medium_sale.append(sale)
  else:
    large_sale.append(sale)

print('Small Sale: ', small_sale)
print('Medium Sale: ', medium_sale)
print('Large Sale: ', large_sale)

```

```python
flavour_ice_cream = ['Vanila', 'Chocolate', 'Strawberry', 'Mango']

i = 0

for flavour in flavour_ice_cream:
  print(i, flavour)
  i += 1
```

```python
cust_dict = {'Name': 'John Smith', 'Amount Spent': 100, 'Items Purchased': 2, 'Revenue': 'Yes'}

for item in cust_dict:
  print(item)
```

```python
for item in cust_dict.values():
  print(item)
```

```python
for item in cust_dict.items():
  print(item)
```

```python
for key, value in cust_dict.items():
  print(key, '->', value)
```

```python
div_by_three = []

for num in range(100):
  if num % 3 == 0:
    div_by_three.append(num)

print(div_by_three)
```

```python
# You're given a list of test scores and your job is to label them "Good Grade" or "Bad Grade".

scores = [85, 90, 78, 92, 88, 75, 98, 60, 82, 95]

for score in scores:
  if score >= 75:
    print(score, ' is Good Grade')
  else:
    print(score, ' is Bad Grade')
```

### Nested For Loops





```python
greetings = ['Hey', 'Howdy', 'Sup', 'Hello']

x = 0
y = 0

for word in greetings:
  for letter in word:
    print(letter)
```

```python
greetings = ['Hey', 'Howdy', 'Sup', 'Hello']

x = 0
y = 0

for word in greetings:
  x += 1
  for letter in word:
    print(x, letter, y)
    y += 1
```

```python
flavours = ['Vanilla', 'Chocolate', 'Strawberry']
toppings = ['Hot Fudge', 'Whipped Cream', 'Oreos']

for flavour in flavours:
  for topping in toppings:
    print(flavour, 'topped with', topping)
```

```python
test_match_venue = ['Lords', 'Melbourne', 'Kolkata', 'Sydney', 'Auckland']
test_match_player = ['Sachin', 'Saurav', 'Dravid', 'Laxman', 'Sehwag']

for venue in test_match_venue:
  for player in test_match_player:
    print(player, 'played in the ', venue, ' cricket ground!')
```
