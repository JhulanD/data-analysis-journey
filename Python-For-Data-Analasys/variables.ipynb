🐍 Python Basics — Variable Naming Conventions
🔹 Use Descriptive Names
Variable names should clearly indicate the kind of data they hold.

✅ age is better than a
✅ employee_salary is better than es
🔹 Use Lowercase Letters and Underscores
Follow Python’s official style guide (PEP 8): use snake_case.

✅ employee_count
✅ total_salary
🔹 Avoid Using Python Keywords
Python keywords are reserved words with special meanings. Do not use them as variable names. Examples of keywords: for, while, if, else, import, break, continue.

🔹 Keep Names Short, But Not Too Short
Names should be easy to type but still meaningful.

✅ num → good for a number
❌ n → too cryptic
❌ number_of_items_in_inventory → too long
🔹 Avoid Single Character Names
Except for common cases like i for loop indices or x, y, z for coordinates, avoid single characters. They are not descriptive enough.

🔹 Use English Words
Even if English isn’t your first language, using English words makes your code more accessible globally.

✅ customer_age
❌ kund_age (localized word)
🔹 Be Consistent
Stick to the naming convention used in your project.

If the project uses snake_case, follow it.
If it uses camelCase, stay consistent.
✅ Following these conventions makes your code cleaner, easier to read, and more professional.

````python
# Use Descriptive Names

# Bad
a = 25
es = 50000

# Good
age = 25
employee_salary = 50000

print(age)              # 25
print(employee_salary)  # 50000
````

### The Core Concept: How Indexing Works
To slice a string, you first need to understand how Python numbers each character position (indexing). Python counts positions starting from **0** from left to right, but it can also count backward starting from **-1** from right to left.

For the word `"PYTHON"`:
* **Positive Index (Left to Right):** `P=0`, `Y=1`, `T=2`, `H=3`, `O=4`, `N=5`
* **Negative Index (Right to Left):** `N=-1`, `O=-2`, `H=-3`, `T=-4`, `Y=-5`, `P=-6`

---

### The Basic Syntax
The standard formula for slicing is:
```python
string[start : stop : step]
```
* **`start`**: The index where you want to begin cutting (**inclusive**—this character is kept).
* **`stop`**: The index where you want to finish cutting (**exclusive**—the slice stops *just before* this character).
* **`step`**: (Optional) Dictates the spacing. A step of `2` takes every second character.

---

### Code Examples

#### 1. Basic Slice (Cutting out a middle chunk)
```python
text = "PythonProgramming"
# Extract "Python" (Indices 0 to 5. Stops before index 6)
print(text[0:6])  # Output: Python
```

#### 2. Omitting Start or Stop Shortcuts
If you leave a side blank, Python assumes you mean the very beginning or the absolute end.
```python
text = "PythonProgramming"

# Leave 'start' blank to go from the beginning up to index 6
print(text[:6])   # Output: Python

# Leave 'stop' blank to go from index 6 all the way to the end
print(text[6:])   # Output: Programming
```

#### 3. Using Negative Indexes
Perfect for when you want to target characters relative to the end of the text string.
```python
filename = "report_2026.pdf"
# Get the last 3 characters
print(filename[-3:])  # Output: pdf
```

#### 4. Skipping Characters and Reversing
```python
numbers = "12345678"
# Get every second character
print(numbers[::2])   # Output: 1357

# Reverse a string quickly using a step of -1
print(numbers[::-1])  # Output: 87654321
```

---

### Real-Life Use Cases

#### Case A: Cleaning Up User Phone Numbers
When users type their phone numbers, you might only want to store the core 10-digit number and strip away an added country code.
```python
user_input = "+15551234567"
# Skip the country code (+1) and grab the rest
clean_number = user_input[2:]
print(clean_number)  # Output: 5551234567
```

#### Case B: Extracting Dates from Log Files
System log strings usually bundle dates right at the beginning of a long text line.
```python
log_line = "2026-07-27 ERROR: Database connection failed."
# Extract just the date component (the first 10 characters)
log_date = log_line[:10]
print(log_date)  # Output: 2026-07-27
```

#### Case C: Hiding Credit Card Details (Masking)
For security, checkout screens mask credit card numbers, revealing only the final digits.
```python
card_number = "1234567890123456"
# Grab the last 4 characters and attach them to mask symbols
masked_card = "**** **** **** " + card_number[-4:]
print(masked_card)  # Output: **** **** **** 3456
```


````python
# 🐍 String Slicing in Python

# Slicing allows us to extract parts of a string using the syntax:
# string[start:stop:step]
# - start → index where slice begins (default: 0)
# - stop  → index where slice ends (not included, default: end of string)
# - step  → increment between indices (default: 1)

text = "DataEngineering"

# 1️⃣ Basic slicing
print(text[0:4])   # 'Data' → characters from index 0 up to (but not including) 4
print(text[4:])    # 'Engineering' → from index 4 to the end
print(text[:4])    # 'Data' → from start up to index 4

# 2️⃣ Negative indices
# Negative indices count from the end of the string
print(text[-3:])   # 'ing' → last 3 characters
print(text[:-3])   # 'DataEngineer' → everything except last 3 characters

# 3️⃣ Step parameter
numbers = "0123456789"

print(numbers[::2])   # '02468' → every 2nd character
print(numbers[1::2])  # '13579' → every 2nd character starting from index 1
print(numbers[::-1])  # '9876543210' → reverse the string

# 4️⃣ Combining start, stop, and step
print(numbers[2:8:2]) # '246' → from index 2 to 7, step of 2

# 5️⃣ Practical examples
sentence = "PythonRocks"

# Extract 'Python'
print(sentence[0:6])   # 'Python'

# Extract 'Rocks'
print(sentence[6:])    # 'Rocks'

# Reverse the string
print(sentence[::-1])  # 'skcoRnothyP'

# Extract every second character
print(sentence[::2])   # 'PtoRcs'

# 6️⃣ Edge cases
print(sentence[:])     # 'PythonRocks' → full string (default start & stop)
print(sentence[::])    # 'PythonRocks' → same as above
print(sentence[100:])  # '' → slicing beyond length returns empty string
````
