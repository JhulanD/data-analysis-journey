# Comprehensive Guide to Python Data Types

Python data types categorize the kind of value a variable holds. This classification determines what operations you can safely perform on that value.

Python is a **dynamically typed** language. This means you do not need to explicitly declare a variable's type when you create it. The Python interpreter automatically detects and assigns the type at runtime based on the value provided. You can check the data type of any object using the built-in `type()` function.

---

## 1. Numeric Types

Numeric types represent numerical data. They are used for calculations, counts, measurements, and tracking precise values.

### `int` (Integer)
* **Description:** Represents positive or negative whole numbers without decimals. In Python, integers have arbitrary precision, meaning they can grow as large as your system's memory allows.
* **Examples:** `10`, `-500`, `123456789012345`

### `float` (Floating-Point Number)
* **Description:** Represents real numbers containing a decimal point. Floats can also represent scientific notation using the character `e` or `E` to signify powers of 10.
* **Examples:** `3.14`, `-0.001`, `2.5e2` (which equals `250.0`)

### `complex` (Complex Number)
* **Description:** Represents numbers with a real part and an imaginary part. The imaginary part is written with a `j` or `J` suffix.
* **Examples:** `3 + 4j`, `-1j`

```python
# Numeric Examples
age = 25          # int
price = 19.99     # float
coords = 2 + 3j   # complex

print(type(age))    # <class 'int'>
print(type(price))  # <class 'float'>
```

---

## 2. Text & Sequence Types

Sequence types represent ordered collections of items. Items within a sequence are indexed using integer positions starting from `0`.

### `str` (String)
* **Description:** Represents textual data. Strings are collections of characters wrapped in single (`'...'`), double (`"..."`), or triple (`'''...'''` or `"""..."""`) quotes. Triple quotes allow strings to span multiple lines.
* **Examples:** `"Hello"`, `'Python'`, `"""Multi-line string"""`

### `list` (List)
* **Description:** Represents an ordered, changeable (mutable) collection of items. Lists can contain duplicate values and elements of differing data types. They are defined using square brackets `[]`.
* **Examples:** `["apple", 42, True, 3.14]`

### `tuple` (Tuple)
* **Description:** Represents an ordered, unchangeable (immutable) collection of items. Like lists, tuples can hold diverse data types and duplicates. They are defined using parentheses `()`.
* **Examples:** `(10, 20)`, `("London", "UK")`

### `range` (Range)
* **Description:** Represents an immutable sequence of numbers. It is commonly used for looping a specific number of times in `for` loops. It is generated using the `range(start, stop, step)` function.
* **Examples:** `range(5)` (generates `0, 1, 2, 3, 4`)

```python
# Sequence Examples
message = "Learning Python"       # str
shopping_list = ["milk", "eggs"]  # list
point = (4.5, 7.2)                # tuple
numbers = range(1, 10)            # range
```

---

## 3. Mapping Type

Mapping types store data pairs consisting of a unique key and an associated value.

### `dict` (Dictionary)
* **Description:** Represents an unordered (ordered by insertion since Python 3.7), changeable collection of key-value pairs. Keys must be unique and immutable data types (like strings or integers), while values can be of any type. Dictionaries are defined using curly braces `{}` with colons separating keys and values.
* **Examples:** `{"name": "Alice", "age": 30}`, `{1: "One", 2: "Two"}`

```python
# Mapping Example
user_profile = {
    "username": "coder123",
    "followers": 1500,
    "is_active": True
}
print(user_profile["username"])  # Output: coder123
```

---

## 4. Set Types

Set types represent collections of unique, unordered items. They are ideal for eliminating duplicates and performing mathematical set operations like unions or intersections.

### `set` (Set)
* **Description:** Represents an unordered, mutable collection of unique items. Sets do not allow duplicate values. Because they are unordered, elements cannot be accessed via an index. They are defined using curly braces `{}` or the `set()` constructor.
* **Examples:** `{1, 2, 3}`, `{"apple", "banana"}`

### `frozenset` (Frozen Set)
* **Description:** Represents an immutable version of a set. Once created, elements cannot be added or removed.
* **Examples:** `frozenset([1, 2, 3])`

```python
# Set Example
unique_numbers = {1, 2, 2, 3, 4}
print(unique_numbers)  # Output: {1, 2, 3, 4} (duplicates automatically removed)
```

---

## 5. Boolean & None Types

These types handle logical operations, conditional states, and empty placeholders.

### `bool` (Boolean)
* **Description:** Evaluates to one of exactly two values: `True` or `False`. Booleans result from comparison operations or logical expressions.
* **Examples:** `True`, `False`

### `NoneType` (None)
* **Description:** Represents the intentional absence of a value or a null pointer. It is denoted by the keyword `None`. Functions that do not explicitly return a value will return `None` by default.
* **Examples:** `None`

```python
# Boolean and None Examples
is_logged_in = False
empty_seat = None
```

## Core Concept: Mutability vs. Immutability

Every object in Python is classified as either **mutable** (can be changed after creation) or **immutable** (cannot be changed after creation).

| Mutability Class | Data Types | Can It Be Modified In Place? | Memory Behavior |
| :--- | :--- | :--- | :--- |
| **Mutable** | `list`, `dict`, `set` | **Yes** | Items can be added, updated, or deleted without changing the object's identity. |
| **Immutable** | `int`, `float`, `str`, `tuple`, `bool`, `frozenset` | **No** | Modifying the value creates a completely new object in memory. |

# Sets


A set in Python is an unordered collection of unique items. Sets are defined by enclosing a comma-separated sequence of items in curly braces {}.


Note that sets are unordered, which means that the order in which items are added to a set is not necessarily the same order in which they are iterated.

Unlike lists and tuples, sets cannot contain duplicate items. If you try to add a duplicate item to a set, it will be ignored.

Sets are mutable, which means you can add and remove items from a set. However, because sets are unordered, they do not support indexing or slicing.

### Creating Sets
Sets can be created using curly braces `{}` or the `set()` constructor. Remember, sets only store unique elements, and duplicates are automatically removed.

````python
my_set = {1, 2, 3, 2, 1}
print(f"Set created with duplicates: {my_set}")

another_set = set([4, 5, 5, 6])
print(f"Set created from a list: {another_set}")

# An empty set must be created using set(), not {} (which creates an empty dictionary)
empty_set = set()
print(f"Empty set: {empty_set}")
````

### Adding and Removing Elements
You can add elements to a set using the `.add()` method and remove them using `.remove()` or `.discard()`. `.remove()` raises an error if the element is not found, while `.discard()` does not.

````python
my_set.add(4)
print(f"Set after adding 4: {my_set}")

my_set.remove(2)
print(f"Set after removing 2: {my_set}")

my_set.discard(10) # No error if 10 is not present
print(f"Set after discarding 10: {my_set}")

# my_set.remove(10) # Uncommenting this line would raise a KeyError
````

### Set Operations
Sets are powerful for performing mathematical set operations like union, intersection, difference, and symmetric difference.

````python
set_a = {1, 2, 3, 4}
set_b = {3, 4, 5, 6}

# Union: All unique elements from both sets
union_set = set_a.union(set_b)
print(f"Union (set_a | set_b): {union_set}")

# Intersection: Elements common to both sets
intersection_set = set_a.intersection(set_b)
print(f"Intersection (set_a & set_b): {intersection_set}")

# Difference: Elements in set_a but not in set_b
difference_set = set_a.difference(set_b)
print(f"Difference (set_a - set_b): {difference_set}")

# Symmetric Difference: Elements in either set_a or set_b, but not in both
symmetric_difference_set = set_a.symmetric_difference(set_b)
print(f"Symmetric Difference (set_a ^ set_b): {symmetric_difference_set}")

# Check for subsets and supersets
print(f"Is {{1, 2}} a subset of set_a? {{1, 2}}.issubset(set_a): {{1, 2}}.issubset(set_a)")
print(f"Is set_a a superset of {{1, 2}}? set_a.issuperset({{1, 2}}): {set_a.issuperset({1, 2})}")
````

### Use Case: Removing Duplicates from a List
One common use of sets is to efficiently remove duplicate elements from a list.

````python
data_with_duplicates = [1, 5, 2, 8, 2, 1, 9, 5]
print(f"Original list with duplicates: {data_with_duplicates}")

unique_data = list(set(data_with_duplicates))
print(f"List after removing duplicates using a set: {unique_data}")
````

## Dictionaries in Python
A dictionary in Python is an unordered collection of items where each item is stored as a key-value pair. You use a unique key to access its associated value.

**Key Characteristics:**

Definition: Dictionaries are defined by enclosing a comma-separated sequence of key-value pairs in curly braces {}. The key and value are separated by a colon :. For example: person = {'name': 'Alice', 'age': 25, 'city': 'New York'}
Mutability: Dictionaries are mutable, meaning you can add, remove, or change items (key-value pairs) after the dictionary is created.
Unordered (by index): While dictionaries preserve insertion order since Python 3.7, you cannot access elements by numerical index like you would with lists or tuples. You access elements using their keys.
Keys: Keys must be unique and immutable data types (e.g., strings, integers, tuples). Values can be of any data type.
Common Methods: Python provides methods like .keys() to get all keys, .values() to get all values, and .items() to get all key-value pairs.
Accessing Values: You can access values using the key in square brackets, e.g., ice_cream_customer['Name'].
Dictionaries are incredibly useful for storing structured data where you need to associate specific pieces of information (values) with unique identifiers (keys).

### 9. Converting Data Types (Type Casting)
Type casting allows you to convert a variable from one data type to another. This is often necessary when you need to perform operations that require specific data types, or when cleaning and preparing data. Python provides built-in functions like `int()`, `float()`, `str()`, `list()`, and `set()` for explicit type conversion.

````python
# Converting to Integer (int())
# You can convert numbers represented as strings or other numeric types to an integer.

string_num = "123"
int_from_string = int(string_num)
print(f"'{string_num}' (str) converted to int: {int_from_string}, type: {type(int_from_string)}")

float_num = 45.67
int_from_float = int(float_num) # Note: float is truncated, not rounded
print(f"'{float_num}' (float) converted to int: {int_from_float}, type: {type(int_from_float)}")

# Caution: int() cannot convert a string with decimals directly
# try_int_float_str = int("12.3") # This would raise a ValueError
````

````python
# Converting to Float (float())
# You can convert integers or strings representing numbers to a float.

int_val = 100
float_from_int = float(int_val)
print(f"'{int_val}' (int) converted to float: {float_from_int}, type: {type(float_from_int)}")

string_float = "3.14159"
float_from_string = float(string_float)
print(f"'{string_float}' (str) converted to float: {float_from_string}, type: {type(float_from_string)}")
````

````python
# Converting to String (str())
# Any data type can generally be converted to its string representation.

number = 12345
str_from_int = str(number)
print(f"'{number}' (int) converted to str: '{str_from_int}', type: {type(str_from_int)}")

boolean_val = True
str_from_bool = str(boolean_val)
print(f"'{boolean_val}' (bool) converted to str: '{str_from_bool}', type: {type(str_from_bool)}")

my_list = [1, 2, 3]
str_from_list = str(my_list)
print(f"'{my_list}' (list) converted to str: '{str_from_list}', type: {type(str_from_list)}")
````

````python
# Converting to List (list())
# You can convert iterable objects (like strings, tuples, or sets) into a list.

my_string = "Python"
list_from_string = list(my_string)
print(f"'{my_string}' (str) converted to list: {list_from_string}, type: {type(list_from_string)}")

my_tuple = (10, 20, 30)
list_from_tuple = list(my_tuple)
print(f"'{my_tuple}' (tuple) converted to list: {list_from_tuple}, type: {type(list_from_tuple)}")

my_set = {1, 2, 3}
list_from_set = list(my_set)
print(f"'{my_set}' (set) converted to list: {list_from_set}, type: {type(list_from_set)}")
````

````python
# Converting to Set (set())
# You can convert iterable objects (like lists, tuples, or strings) into a set.
# Remember, sets automatically remove duplicate elements and are unordered.

numbers_list = [1, 2, 2, 3, 4, 4, 5]
set_from_list = set(numbers_list)
print(f"'{numbers_list}' (list) converted to set: {set_from_list}, type: {type(set_from_list)}")

text_string = "hello world"
set_from_string = set(text_string) # Unique characters
print(f"'{text_string}' (str) converted to set: {set_from_string}, type: {type(set_from_string)}")
````

````python
my_list = [1, 2, 3, 4, 5]
my_tuple = tuple(my_list)
print(f"'{my_list}' (list) converted to tuple: {my_tuple}, type: {type(my_tuple)}")
````
